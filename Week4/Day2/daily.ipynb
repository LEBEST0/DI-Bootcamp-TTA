```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# ==========================================
# 1. EXPLORATION ET CHARGEMENT DES DONNÉES
# ==========================================

# Chargement du jeu de données fourni (ex2data1.txt)
# Le fichier contient deux notes d'examens suivies de la cible (0: Refusé, 1: Admis)
try:
    df = pd.read_csv('ex2data1.txt', header=None, names=['Note_Examen_1', 'Note_Examen_2', 'Admission'])
except FileNotFoundError:
    # Option de secours si vous exécutez le code directement en copiant le texte brut
    import io
    data_str = """34.62365962451697,78.0246928153624,0
30.28671076822607,43.89499752400101,0
35.84740876993872,72.90219802708364,0
60.18259938620976,86.30855209546826,1
79.0327360507101,75.3443764369103,1""" # Remplacer par le contenu complet si besoin
    df = pd.read_csv(io.StringIO(data_str), header=None, names=['Note_Examen_1', 'Note_Examen_2', 'Admission'])

# Affichage des premières lignes pour vérifier la structure
print("Aperçu des premières lignes du jeu de données :")
print(df.head())
print(f"\nDimensions du jeu de données : {df.shape}")

# Séparation des classes pour la visualisation
admis = df[df['Admission'] == 1]
refuses = df[df['Admission'] == 0]

# Création du nuage de points (Scatter Plot)
plt.figure(figsize=(8, 6))
plt.scatter(admis['Note_Examen_1'], admis['Note_Examen_2'], color='green', marker='+', s=100, label='Admis (1)')
plt.scatter(refuses['Note_Examen_1'], refuses['Note_Examen_2'], color='red', marker='o', alpha=0.7, label='Non admis (0)')

plt.xlabel('Note à l\'Examen 1')
plt.ylabel('Note à l\'Examen 2')
plt.title('Visualisation des données d\'admission en fonction des notes d\'examens')
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.show()


# ==========================================
# 2. ENTRAÎNEMENT DE LA RÉGRESSION LOGISTIQUE
# ==========================================

# Préparation des caractéristiques (X) et de la variable cible (y)
X = df[['Note_Examen_1', 'Note_Examen_2']]
y = df['Admission']

# Initialisation et entraînement du modèle de régression logistique
model = LogisticRegression()
model.fit(X, y)

# Extraction des meilleurs paramètres trouvés par Scikit-Learn
w = model.coef_[0]  # Poids associés aux deux notes d'examens
b = model.intercept_[0]  # Biais (intercept)

print("\n--- Paramètres du modèle entraîné ---")
print(f"Poids (w) pour l'Examen 1 et l'Examen 2 : {w}")
print(f"Biais (b) : {b}")


# ==========================================
# 3. PRÉDICTIONS ET CALCUL DE LA PRÉCISION
# ==========================================

# Effectuer des prédictions sur l'ensemble du jeu de données
predictions = model.predict(X)

# Calcul du score d'exactitude (Accuracy)
exactitude = accuracy_score(y, predictions)
print(f"\nPrécision (Accuracy) globale du modèle : {exactitude * 100:.2f} %")


# ==========================================
# 4. ÉVALUATION ET FRONTIÈRE DE DÉCISION
# ==========================================

# Visualisation finale avec la frontière de décision linéaire
plt.figure(figsize=(8, 6))
plt.scatter(admis['Note_Examen_1'], admis['Note_Examen_2'], color='green', marker='+', s=100, label='Admis (1)')
plt.scatter(refuses['Note_Examen_1'], refuses['Note_Examen_2'], color='red', marker='o', alpha=0.7, label='Non admis (0)')

# Calcul de la ligne de décision (w1*x1 + w2*x2 + b = 0  =>  x2 = -(w1*x1 + b) / w2)
x1_vals = np.linspace(df['Note_Examen_1'].min(), df['Note_Examen_1'].max(), 100)
x2_vals = -(w[0] * x1_vals + b) / w[1]

plt.plot(x1_vals, x2_vals, color='blue', linestyle='--', linewidth=2, label='Frontière de décision')
plt.xlabel('Note à l\'Examen 1')
plt.ylabel('Note à l\'Examen 2')
plt.title(f'Modèle de Régression Logistique - Précision : {exactitude * 100:.2f}%')
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.show()
```
