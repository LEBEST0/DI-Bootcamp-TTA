```python
import numpy as np
```

```python
from sklearn.linear_model import LogisticRegression

# Example Data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 1, 1, 1])

# Model Training
log_model = LogisticRegression()
log_model.fit(X, y)
predictions = log_model.predict(X)
print("Predictions:", predictions)
```
