```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

Choix du modèle
Pour un problème de prédiction de crédit (classification binaire : défaut vs non-défaut), le choix se porte sur les modèles suivants :

Modèle de référence (Baseline) : La Régression Logistique, pour sa simplicité, sa rapidité d'exécution et sa forte interprétabilité (très importante pour expliquer un refus de prêt aux autorités de régulation).

Modèles avancés : Les algorithmes de boosting d'arbres de décision comme XGBoost ou LightGBM. Ils gèrent nativement les valeurs manquantes, capturent les relations non linéaires complexes et sont extrêmement performants sur les données tabulaires financières.

2. Étapes d'évaluation et indicateurs pertinents
L'évaluation se déroule selon le protocole suivant :

Division des données : Séparation en ensembles d'entraînement (Train) et de test (Test) (ou via une validation croisée stratifiée pour maintenir la proportion des classes).

Gestion du déséquilibre des classes : Les cas de défaut étant minoritaires dans la réalité, l'utilisation de techniques comme le sous-échantillonnage, le sur-échantillonnage (SMOTE) ou l'ajustement des poids des classes est obligatoire.

Indicateurs clés (Metrics) :

Le Rappel (Recall / Sensibilité) : La métrique la plus critique. Elle mesure la capacité du modèle à détecter tous les défauts réels. Rater un défaut (Faux Négatif) coûte extrêmement cher à la banque.

La Précision : Mesure la proportion de vrais défauts parmi ceux prédits par le modèle. Une précision trop basse (Faux Positifs) amène la banque à refuser des prêts à de bons clients, entraînant un manque à gagner.

Le Score F1 : La moyenne harmonique entre Précision et Rappel, idéale pour évaluer le modèle global en situation de déséquilibre de classes.

L'AUC-ROC : Évalue la capacité du modèle à classer un emprunteur en défaut au-dessus d'un emprunteur sain, indépendamment du seuil de décision choisi.
