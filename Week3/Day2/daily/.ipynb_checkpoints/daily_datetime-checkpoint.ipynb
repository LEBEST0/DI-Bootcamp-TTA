# Daily Challenge — Working with Dates and Times

## 1. Handling Dates in NumPy

```python
import numpy as np

dates = np.arange('2020-01-01', '2020-02-01', dtype='datetime64[D]')
print('January 2020 dates:\n', dates)
```

## 2. Converting Between Date Formats

```python
formatted_dates = np.datetime_as_string(dates, unit='D')
print('Formatted dates:\n', formatted_dates)
```

## 3. Rolling Statistics and Window Functions

```python
import pandas as pd

ts_data = pd.Series(
    np.random.randn(100),
    index=pd.date_range('2024-01-01', periods=100)
)

rolling_mean = ts_data.rolling(window=5).mean()
window_sum   = ts_data.rolling(window=5).sum()

print('Rolling mean (5 jours) :\n', rolling_mean.head(10))
print('\nWindow sum  (5 jours) :\n', window_sum.head(10))
```

## 4. Simple Moving Average

```python
sma = ts_data.rolling(window=5).mean()
print('Simple moving average :\n', sma.head(10))
```

## Exercise 1 — Data Creation

```python
import numpy as np
import pandas as pd

np.random.seed(0)

dates       = pd.date_range('2022-01-01', '2022-12-31', freq='D')
temperature = np.random.uniform(10, 35, size=len(dates))
humidity    = np.random.uniform(30, 90, size=len(dates))

df = pd.DataFrame({'Date': dates, 'Temperature': temperature, 'Humidity': humidity})
df.set_index('Date', inplace=True)

print(df.head())
print(f'\nShape : {df.shape}')
```

## Exercise 2 — Data Analysis

```python
# Statistiques descriptives
print('--- Temperature ---')
print(f'Moyenne   : {df["Temperature"].mean():.2f} C')
print(f'Mediane   : {df["Temperature"].median():.2f} C')
print(f'Ecart-type: {df["Temperature"].std():.2f} C')

print('\n--- Humidite ---')
print(f'Moyenne   : {df["Humidity"].mean():.2f} %')
print(f'Mediane   : {df["Humidity"].median():.2f} %')
print(f'Ecart-type: {df["Humidity"].std():.2f} %')

# Dates extremes
print(f'\nJour le plus chaud  : {df["Temperature"].idxmax().date()}  ({df["Temperature"].max():.2f} C)')
print(f'Jour le plus humide : {df["Humidity"].idxmax().date()}  ({df["Humidity"].max():.2f} %)')
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)

axes[0].plot(df.index, df['Temperature'], color='tomato', linewidth=0.8, alpha=0.8)
axes[0].set_ylabel('Temperature (C)')
axes[0].set_title('Temperature et humidite en 2022')
axes[0].grid(True, alpha=0.3)

axes[1].plot(df.index, df['Humidity'], color='steelblue', linewidth=0.8, alpha=0.8)
axes[1].set_ylabel('Humidite (%)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Exercise 3 — Window Functions (7 jours)

```python
df['Temp_MA7']     = df['Temperature'].rolling(window=7).mean()
df['Humidity_MA7'] = df['Humidity'].rolling(window=7).mean()

print(df[['Temperature','Temp_MA7','Humidity','Humidity_MA7']].head(10))
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)

axes[0].plot(df.index, df['Temperature'], color='tomato',   linewidth=0.6, alpha=0.5, label='Brut')
axes[0].plot(df.index, df['Temp_MA7'],    color='darkred',  linewidth=1.8, label='MA 7 jours')
axes[0].set_ylabel('Temperature (C)')
axes[0].set_title('Temperature — donnees brutes vs moyenne mobile 7 jours')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(df.index, df['Humidity'],    color='steelblue', linewidth=0.6, alpha=0.5, label='Brut')
axes[1].plot(df.index, df['Humidity_MA7'],color='darkblue',  linewidth=1.8, label='MA 7 jours')
axes[1].set_ylabel('Humidite (%)')
axes[1].set_title('Humidite — donnees brutes vs moyenne mobile 7 jours')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
