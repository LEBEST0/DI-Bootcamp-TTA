# Initial Data Exploration

```python
# importation of libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("Apple Stock Prices.csv")
```

```python
data.head(15)

```

```python
data.isna().sum()
```

```python
data.info()
```

```python
# Traitement
```

```python
data["Date"] = pd.to_datetime(data["Date"], format="%d/%m/%Y")
```

```python
data.head()
```

```python
data = data.sort_values("Date")
data.set_index("Date", inplace=True)
```

```python
# Analyse descriptive
```

```python
# Evaluons les tendences centrales

moyenne=data["Adj Close"].mean()
mini=data["Adj Close"].min()
maxi=data["Adj Close"].max()
print(f"La moyenne des couts est de: {moyenne} de 1981 à 2023" )
print(f"La valeur minimale de coût de l'action: {mini} de 1981 à 2023" )
print(f"La valeur maximale de coût de l'action: {maxi} de 1981 à 2023" )
```

```python
data.head(40)
```

```python
plt.plot(data.index,data["Adj Close"])
plt.show()
```

```python
ligne=data.index
inter_data = data["Adj Close"].groupby(data.index.year).mean()
inter_data_max = data["Adj Close"].groupby(data.index.year).max()

```

```python
plt.plot(inter_data)
plt.plot(inter_data_max)
plt.show()
```

in the first year, the Adjusted price was almost insignificant : from 1981 to 2005 around. But between 2005 and 2010, the adjusted price started to growth until the max that's 180.959747 in .

```python
import pandas as pd
import plotly.graph_objects as go

# 1. Création des données (identiques à notre exemple précédent)
data = {
    "Date": ["2026-06-03"],
    "Ouverture": [100],
    "Plus_Haut": [170],
    "Plus_Bas": [90],
    "Cloture": [150],
}

df = pd.DataFrame(data)

# 2. Construction du graphique Candlestick
fig = go.Figure(
    data=[
        go.Candlestick(
            x=df["Date"],
            open=df["Ouverture"],
            high=df["Plus_Haut"],
            low=df["Plus_Bas"],
            close=df["Cloture"],
            increasing_line_color="green",  # Couleur si hausse
            decreasing_line_color="red",  # Couleur si baisse
        )
    ]
)

# 3. Personnalisation des titres et options
fig.update_layout(
    title="Exemple de Candlestick avec Plotly",
    yaxis_title="Prix (€)",
    xaxis_title="Date",
    xaxis_rangeslider_visible=False,  # Désactive le curseur du bas pour simplifier
)

# 4. Affichage du graphique
fig.show()
```
