```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# 1. Data Preparation

```python
temp=np.random.uniform(-5, 35, (10,12))
```

```python
month=np.array([
    "Janvier", "Février", "Mars", "Avril", "Mai", "Juin", 
    "Juillet", "Août", "Septembre", "Octobre", "Novembre", "Décembre"
])
```

```python
cities = np.array([
    "Abidjan",
    "Bouaké",
    "Korhogo",
    "Daloa",
    "San-Pédro",
    "Yamoussoukro",
    "Man",
    "Gagnoa",
    "Divo",
    "Soubré"
])
```

```python
data=pd.DataFrame(index=cities,columns=month,data=temp)
```

```python
data
```

# 2. Data Visualization

```python
data
```

```python
# VERSION AMEliore
import matplotlib.pyplot as plt
for city, city_temp in zip(cities, temp):
    plt.plot(month, city_temp, data=data.loc[city], label=city)

plt.legend()
plt.show()
```

# 2. Data Analysis

```python
# Calcul de la température annuelle moyenne par ville
annual_avg = data.mean(axis=1)
print("Température annuelle moyenne par ville (°C) :")
print(annual_avg.round(2))
```

```python
# Ville avec la température moyenne la plus élevée
hottest_city = annual_avg.idxmax()
hottest_temp = annual_avg.max()

# Ville avec la température moyenne la plus basse
coldest_city = annual_avg.idxmin()
coldest_temp = annual_avg.min()

print(f"🌡️  Ville la plus chaude  : {hottest_city} ({hottest_temp:.2f} °C)")
print(f"❄️  Ville la plus froide  : {coldest_city} ({coldest_temp:.2f} °C)")
```

```python
# Visualisation : température annuelle moyenne par ville (bar chart)
fig, ax = plt.subplots(figsize=(10, 5))
colors = ['tomato' if c == hottest_city else ('steelblue' if c == coldest_city else 'mediumseagreen')
          for c in annual_avg.index]
bars = ax.bar(annual_avg.index, annual_avg.values, color=colors, edgecolor='white', linewidth=0.7)
ax.set_title("Température annuelle moyenne par ville", fontsize=14, fontweight='bold')
ax.set_xlabel("Ville")
ax.set_ylabel("Température moyenne (°C)")
ax.axhline(annual_avg.mean(), color='gray', linestyle='--', label=f"Moyenne globale ({annual_avg.mean():.1f} °C)")
ax.legend()
plt.tight_layout()
plt.show()
```

# 3. Rapport de synthèse

## Résumé des résultats

### Données
Un jeu de données synthétique a été généré avec **NumPy** (`np.random.uniform(-5, 35, (10, 12))`), représentant les températures mensuelles moyennes (en °C) de **10 villes ivoiriennes** sur **12 mois**. Ce tableau a été converti en **DataFrame Pandas** avec les villes en index et les mois en colonnes.

### Analyse
- La **température annuelle moyenne** a été calculée pour chaque ville avec `data.mean(axis=1)` (moyenne sur les colonnes = les mois).
- La ville la plus chaude a été identifiée via `idxmax()` et la plus froide via `idxmin()`.

### Résultats clés
| Indicateur | Ville | Température |
|---|---|---|
| 🌡️ Plus chaude | voir sortie ci-dessus | valeur affichée |
| ❄️ Plus froide | voir sortie ci-dessus | valeur affichée |

### Tendances observées
- Les températures **varient considérablement** d'un mois à l'autre au sein d'une même ville, ce qui est attendu avec des données aléatoires uniformes.
- Le graphique en courbes (partie 3) montre que **certaines villes présentent de fortes amplitudes thermiques** tandis que d'autres sont plus stables.
