```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# 1. Data Preparation

```python
temp=np.random.uniform(-5, 35, (10,12))
```

```python
month=np.array([
    "Janvier", "Février", "Mars", "Avril", "Mai", "Juin", 
    "Juillet", "Août", "Septembre", "Octobre", "Novembre", "Décembre"
])
```

```python
cities = np.array([
    "Abidjan",
    "Bouaké",
    "Korhogo",
    "Daloa",
    "San-Pédro",
    "Yamoussoukro",
    "Man",
    "Gagnoa",
    "Divo",
    "Soubré"
])
```

```python
data=pd.DataFrame(index=cities,columns=month,data=temp)
```

```python
data
```

# 2. Data Analysis

```python
moy=data.mean(axis=1)
```

```python

print(f"The city with greatest temperature is { moy.idxmax } ")
print(f"The city with lower temperature is { moy.idxmin() } ")
```

# 3. Data Visualization

```python
data
```

```python
i=0
for citie in cities:
    plt.plot(month,temp[i], data=data.loc[citie])
    i+=1
plt.legend()
```

```python
import matplotlib.pyplot as plt
# Version améliorée
for city, city_temp in zip(cities, temp):
    plt.plot(month, city_temp, data=data.loc[city], label=city)

plt.legend()
plt.show()
```
